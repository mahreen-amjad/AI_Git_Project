```python
import numpy as np

# Given Dataset
X = np.array([1, 2, 3, 4, 5])
Y = np.array([40, 50, 60, 70, 80])
n = len(X)

# Task 1: Calculate Sums
sum_X = np.sum(X)
sum_Y = np.sum(Y)
sum_XY = np.sum(X * Y)
sum_X2 = np.sum(X**2)

print(f"1. Sum of X (sum_X): {sum_X}")
print(f"   Sum of Y (sum_Y): {sum_Y}")
print(f"   Sum of XY (sum_XY): {sum_XY}")
print(f"   Sum of X^2 (sum_X2): {sum_X2}")

# Task 2: Calculate Means
mean_X = np.mean(X)
mean_Y = np.mean(Y)
print(f"\n2. Mean of X (X_bar): {mean_X}")
print(f"   Mean of Y (Y_bar): {mean_Y}")

# Task 3: Calculate Slope (m)
m = (n * sum_XY - sum_X * sum_Y) / (n * sum_X2 - (sum_X)**2)
print(f"\n3. Slope (m): {m}")

# Task 4: Calculate Intercept (c)
c = mean_Y - m * mean_X
print(f"\n4. Intercept (c): {c}")

# Task 5: Linear Regression Equation
print(f"\n5. Final Linear Regression Equation: Y = {m}X + {c}")

# Task 6: Predict marks for 6 hours
hours_new = 6
predicted_marks = m * hours_new + c
print(f"\n6. Predicted Marks for {hours_new} hours of study: {predicted_marks}")

# Task 7: Interpretation
print("\n7. Interpretation of Slope:")
print(f"   A slope of {m} indicates that for every 1-hour increase in study time,")
print("   the predicted marks increase by 10 units.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Given Dataset
X = np.array([1, 2, 3, 4, 5])
Y = np.array([40, 50, 60, 70, 80])
n = len(X)

# Task 1: Calculate Sums
sum_X = np.sum(X)
sum_Y = np.sum(Y)
sum_XY = np.sum(X * Y)
sum_X2 = np.sum(X**2)

print(f"1. Sum of X: {sum_X}")
print(f"   Sum of Y: {sum_Y}")
print(f"   Sum of XY: {sum_XY}")
print(f"   Sum of X^2: {sum_X2}")

# Task 2: Calculate Means
mean_X = np.mean(X)
mean_Y = np.mean(Y)
print(f"\n2. Mean of X: {mean_X}")
print(f"   Mean of Y: {mean_Y}")

# Task 3 & 4: Calculate Slope (m) and Intercept (c)
m = (n * sum_XY - sum_X * sum_Y) / (n * sum_X2 - (sum_X)**2)
c = mean_Y - m * mean_X
print(f"\n3. Slope (m): {m}")
print(f"4. Intercept (c): {c}")

# Task 5: Linear Regression Equation
print(f"\n5. Final Linear Regression Equation: Y = {m}X + {c}")

# Task 6: Predict marks for 6 hours
hours_new = 6
predicted_marks = m * hours_new + c
print(f"\n6. Predicted Marks for {hours_new} hours of study: {predicted_marks}")

# Task 7: Interpretation
print(f"\n7. Interpretation: Slope {m} means with every increase of 1 hour 10 marks also increased.")

# --- Graphical Representation ---
plt.figure(figsize=(8, 5))
# Actual data points
plt.scatter(X, Y, color='blue', s=100, label='Actual Data Points')

# Regression line
X_line = np.linspace(0, 7, 100)
Y_line = m * X_line + c
plt.plot(X_line, Y_line, color='red', linewidth=2, label=f'Regression Line: Y = {int(m)}X + {int(c)}')

# Prediction point for 6 hours
plt.scatter([hours_new], [predicted_marks], color='green', marker='*', s=250, label=f'Prediction (6 hrs): {predicted_marks}')

plt.title('Linear Regression Model: Hours Studied vs Marks Obtained', fontsize=12)
plt.xlabel('Hours Studied (X)', fontsize=10)
plt.ylabel('Marks Obtained (Y)', fontsize=10)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Ignore scikit-learn feature names warnings for cleaner output
warnings.filterwarnings("ignore", category=UserWarning)

# 1. Create the Training Dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Study_Hours": [2, 5, 6, 3, 7],
    "Attendance": [60, 80, 85, 65, 90],
    "Assignment": ["No", "Yes", "Yes", "No", "Yes"],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass"],
}

df = pd.DataFrame(data)

# Preprocessing categorical features into numeric values
df["Assignment_Encoded"] = df["Assignment"].map({"Yes": 1, "No": 0})
df["Result_Encoded"] = df["Result"].map({"Pass": 1, "Fail": 0})

# Define Features (X) and Target (y) as numpy arrays to avoid warnings
X = df[["Study_Hours", "Attendance", "Assignment_Encoded"]].values
y = df["Result_Encoded"].values

# 2. Initialize and Train Random Forest Classifier (n_estimators=3 trees)
rf_model = RandomForestClassifier(n_estimators=3, random_state=42)
rf_model.fit(X, y)

# 3. New Student Data: Study Hours = 5, Attendance = 82%, Assignment = Yes (1)
new_student = np.array([[5, 82, 1]])

# 4. Predict Final Result
prediction = rf_model.predict(new_student)
predicted_label = "Pass" if prediction[0] == 1 else "Fail"

print("--- Random Forest Practical Execution ---")
print(f"Final Random Forest Prediction for New Student: {predicted_label}")

# Collect individual tree outputs for voting analysis
tree_predictions = []
for i, tree in enumerate(rf_model.estimators_):
  tree_pred = tree.predict(new_student)
  label = "Pass" if tree_pred[0] == 1 else "Fail"
  tree_predictions.append(label)
  print(f"Decision Tree {i+1}: {label}")

# --- Graphical Representation (Voting Distribution) ---
pass_votes = tree_predictions.count("Pass")
fail_votes = tree_predictions.count("Fail")

votes_data = {"Classes": ["Pass", "Fail"], "Votes": [pass_votes, fail_votes]}

plt.figure(figsize=(7, 4.5))
bars = plt.bar(
    votes_data["Classes"],
    votes_data["Votes"],
    color=["#2ecc71", "#e74c3c"],
    width=0.5,
)

# Adding values on top of bars
for bar in bars:
  yval = bar.get_height()
  plt.text(
      bar.get_x() + bar.get_width() / 2,
      yval + 0.05,
      int(yval),
      ha="center",
      va="bottom",
      fontsize=12,
      fontweight="bold",
  )

plt.title(
    "Random Forest Majority Voting Distribution (New Student)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Prediction Class", fontsize=10)
plt.ylabel("Number of Votes (Trees)", fontsize=10)
plt.ylim(0, 3.5)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```
